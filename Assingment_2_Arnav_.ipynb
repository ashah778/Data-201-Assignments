Out of State Tuition if Private vs Not Private

```python

# I am creating boxplots for out of state tuition for whether the college is private or not
# I am adding a title which describes which is describing what the boxplot will show
# I am labeling the x-axis as Private
# I am labeling the y-axis as Out of State Tuition

college.boxplot(column='Outstate', by='Private',figsize = (8,6))

plt.title('Out-of-State Tuition by Private Status')

plt.xlabel('Private')
plt.ylabel('Out-of-State Tuition')

plt.show()
```

f)

Creating the Elite Variable and Comparing Tuition

```python
# I am creating the column Elite to show the data for the students that are ranked top 10% in their high school class
# I am counting the amount of colleges that are Elite vs Not Elite
college['Elite'] = pd.cut(
    college['Top10perc'],
    [0, 50, 100],
    labels=['No','Yes']
)

college['Elite'].value_counts()
```

```python
# I am creating side-by-side boxplots of Outstate tuition based on Elite status
college.boxplot(column='Outstate', by='Elite',figsize= (8,6))

# I am adding labels to the graph
plt.xlabel('Elite')
plt.ylabel('Out-of-State Tuition')

# I am displaying the boxplot

plt.show()
```

g)

Histogram

```python

# I am creating four histograms
fig,ax= plt.subplots(2,2,figsize = (13,8))

# This histogram is for the number of applications
college.plot.hist(y = 'Apps', ax = ax[0,0])
ax[0,0].set_title('Applications')
ax[0,0].set_xlabel('Number of Applications')

# This histogram is for the number of student enrollment
college.plot.hist(y = 'Enroll', ax = ax[0,1])
ax[0,1].set_title('Enrollment')
ax[0,1].set_xlabel('Number of Enrolled Students')

# This histogram is for out of state tuition
college.plot.hist(y = 'Outstate', ax = ax[1,0])
ax[1,0].set_title('Out-Of-State Tuition')
ax[1,0].set_xlabel('Tuition (USD)')

# This histogram is for graduation rates
college.plot.hist(y = 'Grad.Rate', ax = ax[1,1])
ax[1,1].set_title('Graduation Rate')
ax[1,1].set_xlabel('Percentage (%)')

plt.tight_layout()

plt.show()




```
